```python

months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

rainfall = [45, 30, 55, 80, 120, 200, 250, 230, 180, 100, 60, 40]
total = sum(rainfall)
average = total / len(rainfall)
position = rainfall.index(max(rainfall))
wettest_month = months[position]
position = rainfall.index(min(rainfall))
driest_month = months[position]
above_avg = [month for month, rain in zip(months, rainfall) if rain > average]
print("Total rainfall:", total)
print("Average rainfall:", average)
print("Wettest month:", wettest_month)
print("Driest month:", driest_month)
print("Months with above-average rainfall:", above_avg)
```
